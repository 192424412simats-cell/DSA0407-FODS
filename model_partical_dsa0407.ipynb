```python

import pandas as pd
import re
from collections import Counter

data = {
    'ID': [1,2,3,4,5,6,7,8],
    'Review': [
        "Great product, works very well",
        "Very poor quality, stopped working in a week",
        "Good value for money, works great",
        "Excellent quality and fast delivery",
        "Not good, poor packaging and poor quality",
        "Works well, good battery and good price",
        "Delivery was late but the product is great",
        "Excellent product, great value"
    ]
}

df = pd.DataFrame(data)

words = []
for review in df['Review']:
    words.extend(re.findall(r'\b\w+\b', review.lower()))

freq = Counter(words)

print("Word Frequency Distribution:")
for word, count in freq.most_common():
    print(word, ":", count)
```

```python

import pandas as pd
import re
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

data = {
    'feedback': [
        "The delivery was fast and the packaging was great!",
        "Customer support is slow, and the app is not easy to use.",
        "I love the product, the quality is excellent.",
        "Great quality but the price is too high.",
        "The app is easy to use and the support team is helpful.",
        "Delivery was late, and the packaging was damaged.",
        "Excellent support, fast delivery, and great price!",
        "The product quality is poor; I am not happy.",
        "Easy to use, great quality, and good price.",
        "Fast delivery and helpful support team."
    ]
}

df = pd.DataFrame(data)
words = []

for text in df['feedback']:
    text = re.sub(r'[^\w\s]', '', text.lower())
    words.extend([w for w in text.split()
                  if w not in ENGLISH_STOP_WORDS])

freq = Counter(words)

n = int(input("Enter N: "))
top = freq.most_common(n)

print("Top N Words:")
for word, count in top:
    print(word, ":", count)

w, c = zip(*top)

plt.bar(w, c)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Customer Feedback Word Analysis")
plt.xticks(rotation=45)
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats

age = [23,23,27,27,39,41,41,47,49,
       50,52,54,54,56,57,58,58,60]

fat = [9.5,26.5,7.8,17.8,31.4,25.9,
       25.9,27.4,27.2,31.2,34.6,42.5,
       28.8,33.4,30.2,34.1,32.9,41.2]

df = pd.DataFrame({
    'Age': age,
    '%Fat': fat
})

print("Mean:\n", df.mean())
print("\nMedian:\n", df.median())
print("\nStandard Deviation:\n", df.std())

df.boxplot()
plt.title("Age and Body Fat Boxplots")
plt.show()

# Scatter plot
plt.scatter(age, fat)
plt.xlabel("Age")
plt.ylabel("%Fat")
plt.title("Age vs Body Fat")
plt.show()

stats.probplot(age, dist="norm", plot=plt)
plt.title("Q-Q Plot of Age")
plt.show()

stats.probplot(fat, dist="norm", plot=plt)
plt.title("Q-Q Plot of Body Fat")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Student': list(range(1,13)),
    'Study Time': [1,2,2.5,3,4,4.5,
                   5,6,6.5,7,8,9],
    'Exam Score': [35,40,45,50,55,58,
                   62,70,72,78,85,92]
}

df = pd.DataFrame(data)

r = df['Study Time'].corr(df['Exam Score'])

print("Correlation Coefficient:", r)

if r > 0:
    print("Positive Correlation")
elif r < 0:
    print("Negative Correlation")
else:
    print("No Correlation")

plt.scatter(df['Study Time'], df['Exam Score'])
plt.xlabel("Study Time (Hours)")
plt.ylabel("Exam Score")
plt.title("Study Time vs Exam Score")
plt.grid()
plt.show()
```
